- `.dtype`

```python
tensor1d = torch.tensor([1,2,3])
print(tensor1d.dtype)
```

```python
floatvec = tensor1d.to(torch.float32)
print(floatvec.dtype)
```

# Logistic Regression Forward Pass

```python
import torch.nn.functional as F
import torch
```

```python
y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.0])
b = torch.tensor([0.0])
```

```python
z = x1 * w1 + b
z
```

```python
a = torch.sigmoid(z)
a
```

```python
loss = F.binary_cross_entropy(a, y)
loss
```

### Computing Gradients via autograd

```python
import torch.nn.functional as F
from torch.autograd import grad
```

```python
y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)
```

```python
z = x1 * w1 + b
```

```python
a = torch.sigmoid(z)
```

```python
loss = F.binary_cross_entropy(a, y)
```

```python
grad_L_w1 = grad(loss, w1, retain_graph=True)
```

```python
grad_L_b = grad(loss, b, retain_graph=True)
```

```python
grad_L_w1
```

```python
grad_L_b
```

```python
# calling backward on loss, pytorch computes the gradients of all the leaf nodes in the graph which will be stored via the tensors' `.grad` attribute

loss.backward()
```

```python
w1.grad
```

```python
b.grad
```
